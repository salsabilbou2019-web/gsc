```python
import pandas as pd
import networkx as nx

file_path = "BIOGRID-MV-Physical-5.0.262.csv"

df = pd.read_csv(
    file_path,
    usecols=[
        "BioGRID ID Interactor A",
        "BioGRID ID Interactor B",
        "Official Symbol Interactor A",
        "Official Symbol Interactor B"
    ],
    dtype=str
)


df = df.dropna(subset=[
    "BioGRID ID Interactor A",
    "BioGRID ID Interactor B"
])
df = df[
    (df["BioGRID ID Interactor A"] != "-") &
    (df["BioGRID ID Interactor B"] != "-") &
    (df["BioGRID ID Interactor A"] !=
     df["BioGRID ID Interactor B"])
].copy()


a = df["BioGRID ID Interactor A"]
b = df["BioGRID ID Interactor B"]
swap = a > b

df["A"] = a.where(~swap, b)
df["B"] = b.where(~swap, a)

edges = df[["A", "B"]].drop_duplicates()

G = nx.from_pandas_edgelist(edges, source="A", target="B")

print("Number of nodes:", G.number_of_nodes())
print("Number of edges:", G.number_of_edges())
```

```python
print("Number of proteins:", G.number_of_nodes())
print("Number of interactions:", G.number_of_edges())
print("Network density:", nx.density(G))
print("Connected components:", nx.number_connected_components(G))
print("Average degree:", sum(dict(G.degree()).values()) / G.number_of_nodes())
```

```python
import matplotlib.pyplot as plt
nodes_100 = list(G.nodes())[:100]
G_sub = G.subgraph(nodes_100)


plt.figure(figsize=(12, 9))
pos = nx.spring_layout(G_sub, seed=42)

nx.draw_networkx(
    G_sub,
    pos,
    with_labels=False,
    node_size=50,
    edge_color="gray",
    alpha=0.7
)

plt.title("BioGRID Protein-Protein Interaction Network (100 proteins)")
plt.axis("off")
plt.show()
```

```python
# Number of connected components
print("Connected components:",
      nx.number_connected_components(G))

# Size of the largest connected component
largest_cc = max(nx.connected_components(G), key=len)
print("Largest component size:", len(largest_cc))

# Visualize the largest connected component
G_largest = G.subgraph(largest_cc)

plt.figure(figsize=(12, 9))
pos = nx.spring_layout(G_largest, seed=42)
nx.draw(G_largest, pos, node_size=20,
        with_labels=False, edge_color="gray")
plt.title("Largest Connected Component of the BioGRID Network")
plt.axis("off")
plt.show()
```

```python
from community import community_louvain
from collections import Counter

partition = community_louvain.best_partition(
    G, random_state=42
)

community_sizes = Counter(partition.values())

print("Number of communities:", len(community_sizes))
print("Ten largest community sizes:",
      community_sizes.most_common(10))
```

```python
largest_community_id = community_sizes.most_common(1)[0][0]

community_nodes = [
    node for node, community_id in partition.items()
    if community_id == largest_community_id
]

H = G.subgraph(community_nodes[:100])

plt.figure(figsize=(9, 7))
pos = nx.spring_layout(H, seed=42)
nx.draw(H, pos, node_size=35, with_labels=False)
plt.title("PPI Network: Sample of a Louvain Community")
plt.show()
```
